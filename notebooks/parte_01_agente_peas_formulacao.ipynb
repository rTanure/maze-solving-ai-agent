## 1. PEAS

### Performance

Funcao sugerida pelo trabalho:

`J = -alpha*custo_caminho - beta*nos_expandidos - gamma*tempo_execucao - delta*movimentos_invalidos - epsilon*celulas_revisitadas`

Preencher no relatorio:

- Quais pesos foram escolhidos.
- Por que custo, tempo, expansoes e revisitas importam neste dominio.
- Se a funcao muda entre busca classica, local e online.

### Environment

- Labirinto discreto em grade.
- Celulas livres, paredes, posicao inicial `A`, objetivo `B`, coletaveis `C` e celulas desconhecidas `?` no modo online.

### Actuators

- Movimentos ortogonais: cima, baixo, esquerda, direita.

### Sensors

- Busca classica: mapa completo conhecido.
- Busca local: mapa completo conhecido e pontos `C` obrigatorios.
- Busca online: percepcao local com raio configuravel, atualizando mapa interno progressivamente.

## 2. Classificacao do agente

Resposta recomendada para desenvolver no relatorio:

- O agente e baseado em objetivos, pois escolhe acoes para atingir `B` e, na busca local, visitar todos os `C` antes de `B`.
- O agente usa modelo interno, pois representa o labirinto como uma matriz e, no modo online, atualiza esse modelo conforme percebe novas celulas.
- Nao e apenas reflexo simples, pois nao decide apenas pela percepcao imediata; ele planeja caminhos.
- Nao e necessariamente aprendiz, pois nao ha mecanismo de aprendizado por experiencia implementado.
- Pode ser analisado como agente baseado em utilidade se a funcao `J` for usada explicitamente para comparar solucoes.

## 3. Formulacao formal - Busca classica

`<S, A, T, s0, G, c>`

- `S`: conjunto de posicoes livres do labirinto, incluindo `A`, `B` e `C` quando `C` nao for objetivo da parte analisada.
- `A`: movimentos ortogonais validos.
- `T`: funcao de transicao que leva de uma celula livre a uma celula livre adjacente.
- `s0`: posicao inicial `A`.
- `G`: teste de objetivo `estado == B`.
- `c`: custo unitario por movimento.

## 4. Formulacao formal - Busca local com coletas

- Estado candidato: uma permutacao dos pontos coletaveis `C`.
- Objetivo: minimizar o custo total da rota `A -> C... -> B`.
- Custo: soma dos menores caminhos entre pontos consecutivos, calculados por UCS ou A*.
- Vizinhanca implementada: troca de dois pontos da ordem de visitacao.

Ponto para discutir: a busca local otimiza a ordem dos coletaveis, mas cada trecho da rota ainda e resolvido por busca classica no labirinto.

## 5. Formulacao formal - Busca online

- O simulador possui o mapa real completo.
- O agente possui apenas um mapa interno parcial.
- Ciclo exigido: perceber -> atualizar mapa interno -> planejar -> agir.
- Objetivo: chegar de `A` ate `B` sem conhecer previamente todo o labirinto.

Observacao importante: pela definicao do trabalho, coletaveis `C` pertencem a Parte III. Na Parte IV, `C` pode ser tratado como celula livre.
